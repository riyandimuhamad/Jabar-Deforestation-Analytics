```python
import pandas as pd

# ==============================================================================
# TAHAP 1: BERSIHKAN & TRANSFORMASI DATA GFW
# ==============================================================================
print("1️⃣ Memproses Data GFW...")

# Filter hanya Jawa Barat
df_gfw_jabar = df_gfw[df_gfw['subnational1'] == 'Jawa Barat'].copy()

# Kamus Perbaikan Nama (Agar sama dengan BNPB)
mapping_nama = {
    'Banjar': 'Kota Banjar',
    'Cimahi': 'Kota Cimahi',
    'Depok': 'Kota Depok'
    # Nama lain seperti 'Bandung', 'Garut' sudah cocok, tidak perlu diubah
}

# Terapkan perbaikan nama
df_gfw_jabar['subnational2'] = df_gfw_jabar['subnational2'].replace(mapping_nama)

# Buang data yang tidak relevan (Waduk Cirata)
df_gfw_jabar = df_gfw_jabar[df_gfw_jabar['subnational2'] != 'Waduk Cirata']

# Lakukan MELT (Ubah kolom tahun tc_loss_ha_20xx menjadi baris)
# Kita cari kolom yang namanya mengandung 'tc_loss_ha_'
kolom_tahun = [col for col in df_gfw_jabar.columns if 'tc_loss_ha_' in col]

df_gfw_clean = df_gfw_jabar.melt(
    id_vars=['subnational2'],       # Kolom identitas (Nama Daerah)
    value_vars=kolom_tahun,         # Kolom-kolom tahun yang mau dilebur
    var_name='Tahun_Raw',           # Nama kolom baru untuk label tahun
    value_name='Luas_Deforestasi'   # Nama kolom baru untuk nilai data
)

# Bersihkan kolom Tahun (dari 'tc_loss_ha_2001' menjadi angka '2001')
df_gfw_clean['Tahun'] = df_gfw_clean['Tahun_Raw'].str.replace('tc_loss_ha_', '').astype(int)

# Rapikan tabel akhir GFW
df_gfw_final = df_gfw_clean[['subnational2', 'Tahun', 'Luas_Deforestasi']].rename(columns={'subnational2': 'Kabupaten'})

print(f"   ✅ Selesai! GFW sekarang punya {len(df_gfw_final)} baris data.")


# ==============================================================================
# TAHAP 2: AGREGASI DATA BNPB
# ==============================================================================
print("\n2️⃣ Memproses Data BNPB...")

# Filter Bencana Hidrometeorologi (Banjir & Longsor)
filter_bencana = df_bnpb['Jenis Bencana'].isin(['Banjir', 'Tanah Longsor', 'Banjir Bandang'])
# Filter Provinsi Jawa Barat (untuk keamanan)
# Pastikan kolom Provinsi dibaca string agar tidak error
df_bnpb['Provinsi'] = df_bnpb['Provinsi'].astype(str)
filter_provinsi = df_bnpb['Provinsi'].str.upper().str.contains('JAWA BARAT')

df_bnpb_jabar = df_bnpb[filter_bencana & filter_provinsi].copy()

# Grouping: Hitung jumlah kejadian per Kabupaten per Tahun
df_bnpb_agg = df_bnpb_jabar.groupby(['Kabupaten', 'Tahun']).size().reset_index(name='Jumlah_Bencana')

print(f"   ✅ Selesai! Data BNPB diringkas menjadi {len(df_bnpb_agg)} baris.")


# ==============================================================================
# TAHAP 3: PENGGABUNGAN (MERGE)
# ==============================================================================
print("\n3️⃣ Menggabungkan Kedua Data...")

# Gabungkan berdasarkan 'Kabupaten' dan 'Tahun'
# Kita gunakan 'inner join' agar hanya mengambil data yang Tahun & Kotanya ada di kedua file
df_final = pd.merge(df_gfw_final, df_bnpb_agg, on=['Kabupaten', 'Tahun'], how='inner')

# Urutkan biar rapi
df_final = df_final.sort_values(by=['Kabupaten', 'Tahun'])

print("\n🎉 PENGGABUNGAN SUKSES!")
print(f"Total Data Siap Analisis: {len(df_final)} baris.")
print("-" * 50)
print(df_final.head(10)) # Tampilkan 10 data pertama

# (Opsional) Simpan ke Excel agar bisa kamu download
df_final.to_excel('dataset_skripsi_final.xlsx', index=False)
print("\n📁 File 'dataset_skripsi_final.xlsx' telah dibuat! Silakan download dari panel kiri.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ==============================================================================
# REVISI TAHAP 1: FILTER THRESHOLD 30% (STANDAR HUTAN)
# ==============================================================================
print("1️⃣ Memproses Ulang GFW (Filter Threshold 30%)...")

# Filter Jawa Barat DAN Threshold 30
# Threshold 30 adalah standar umum untuk mendefinisikan 'Hutan'
df_gfw_jabar = df_gfw[(df_gfw['subnational1'] == 'Jawa Barat') & (df_gfw['threshold'] == 30)].copy()

# Mapping Nama Kota (Sama seperti sebelumnya)
mapping_nama = {
    'Banjar': 'Kota Banjar',
    'Cimahi': 'Kota Cimahi',
    'Depok': 'Kota Depok'
}
df_gfw_jabar['subnational2'] = df_gfw_jabar['subnational2'].replace(mapping_nama)
df_gfw_jabar = df_gfw_jabar[df_gfw_jabar['subnational2'] != 'Waduk Cirata']

# MELT (Melebur Tahun)
kolom_tahun = [col for col in df_gfw_jabar.columns if 'tc_loss_ha_' in col]
df_gfw_clean = df_gfw_jabar.melt(
    id_vars=['subnational2'],
    value_vars=kolom_tahun,
    var_name='Tahun_Raw',
    value_name='Luas_Deforestasi'
)
df_gfw_clean['Tahun'] = df_gfw_clean['Tahun_Raw'].str.replace('tc_loss_ha_', '').astype(int)
df_gfw_final = df_gfw_clean[['subnational2', 'Tahun', 'Luas_Deforestasi']].rename(columns={'subnational2': 'Kabupaten'})

print(f"   ✅ Data GFW Bersih (Tanpa Duplikat): {len(df_gfw_final)} baris.")

# ==============================================================================
# TAHAP 2 & 3: AGREGASI & MERGE (Sama seperti sebelumnya)
# ==============================================================================
# Agregasi BNPB
df_bnpb['Provinsi'] = df_bnpb['Provinsi'].astype(str)
filter_bencana = df_bnpb['Jenis Bencana'].isin(['Banjir', 'Tanah Longsor', 'Banjir Bandang'])
filter_provinsi = df_bnpb['Provinsi'].str.upper().str.contains('JAWA BARAT')
df_bnpb_jabar = df_bnpb[filter_bencana & filter_provinsi].copy()
df_bnpb_agg = df_bnpb_jabar.groupby(['Kabupaten', 'Tahun']).size().reset_index(name='Jumlah_Bencana')

# Merge
df_final = pd.merge(df_gfw_final, df_bnpb_agg, on=['Kabupaten', 'Tahun'], how='inner')
df_final = df_final.sort_values(by=['Kabupaten', 'Tahun'])

# Simpan dataset bersih
df_final.to_excel('dataset_skripsi_FIX_no_duplikat.xlsx', index=False)

print("\n🎉 DATASET FINAL SIAP! (Duplikasi sudah hilang)")
print(f"Total Baris: {len(df_final)}")
print(df_final.head())

# ==============================================================================
# TAHAP 4: UJI COBA VISUALISASI (THE MOMENT OF TRUTH)
# ==============================================================================
print("\n📊 Membuat Grafik Korelasi Sederhana...")

plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_final, x='Luas_Deforestasi', y='Jumlah_Bencana', hue='Kabupaten', alpha=0.6, legend=False)
sns.regplot(data=df_final, x='Luas_Deforestasi', y='Jumlah_Bencana', scatter=False, color='red')

plt.title('Korelasi Deforestasi vs Jumlah Bencana (Banjir/Longsor) di Jawa Barat')
plt.xlabel('Luas Hutan Hilang (Hektar)')
plt.ylabel('Jumlah Kejadian Bencana')
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()

# Cek Nilai Korelasi (Angka Statistik)
korelasi = df_final['Luas_Deforestasi'].corr(df_final['Jumlah_Bencana'])
print(f"\n📈 Nilai Korelasi Pearson: {korelasi:.2f}")
if korelasi > 0:
    print("Kesimpulan Sementara: Positif (+) -> Semakin Hutan Hilang, Bencana Semakin Sering.")
else:
    print("Kesimpulan Sementara: Negatif/Lemah. Perlu analisis lebih dalam.")
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

print("MEMULAI PROSES K-MEANS CLUSTERING...")

# 1. Persiapan Data
# Kita ambil hanya kolom X (Deforestasi) dan Y (Bencana) untuk dicluster
X = df_final[['Luas_Deforestasi', 'Jumlah_Bencana']]

# Standarisasi Data (PENTING untuk K-Means agar skalanya sama)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 2. Menentukan Cluster (Misal kita set 3 Cluster: Aman, Waspada, Bahaya)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_final['Cluster'] = kmeans.fit_predict(X_scaled)

# 3. Analisis Hasil Cluster
# Kita lihat rata-rata Deforestasi & Bencana tiap cluster untuk memberi label
summary_cluster = df_final.groupby('Cluster')[['Luas_Deforestasi', 'Jumlah_Bencana']].mean().reset_index()
summary_cluster = summary_cluster.sort_values(by='Jumlah_Bencana') # Urutkan dari yang bencananya paling sedikit

print("\n=== PROFIL SETIAP CLUSTER (RATA-RATA) ===")
print(summary_cluster)

# Memberi Label Otomatis (Low, Medium, High Risk)
# Cluster dengan rata-rata bencana terendah = 'Aman', tertinggi = 'Bahaya'
label_map = {
    summary_cluster.iloc[0]['Cluster']: 'Rendah (Aman)',
    summary_cluster.iloc[1]['Cluster']: 'Sedang (Waspada)',
    summary_cluster.iloc[2]['Cluster']: 'Tinggi (Bahaya)'
}
df_final['Status_Rawan'] = df_final['Cluster'].map(label_map)

print("\n=== CONTOH HASIL PENGELOMPOKAN WILAYAH ===")
print(df_final[['Kabupaten', 'Tahun', 'Luas_Deforestasi', 'Jumlah_Bencana', 'Status_Rawan']].sample(10))

# 4. Visualisasi Hasil Clustering
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_final,
    x='Luas_Deforestasi',
    y='Jumlah_Bencana',
    hue='Status_Rawan', # Warna beda tiap status
    palette={'Rendah (Aman)': 'green', 'Sedang (Waspada)': 'orange', 'Tinggi (Bahaya)': 'red'},
    s=100, # Ukuran titik
    alpha=0.7
)
plt.title('Pemetaan Zona Rawan Bencana Menggunakan K-Means Clustering')
plt.xlabel('Luas Hutan Hilang (Ha)')
plt.ylabel('Jumlah Kejadian Bencana')
plt.grid(True, linestyle='--', alpha=0.3)
plt.show()

# Simpan lagi file final yang sudah ada Clusternya
df_final.to_excel('dataset_skripsi_FINAL_CLUSTERED.xlsx', index=False)
print("\nFile 'dataset_FINAL_CLUSTERED.xlsx' (dengan label status) siap didownload!")
```

```python
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

print("MENGHITUNG VALIDITAS MODEL (SILHOUETTE SCORE)...")

# 1. Siapkan Data yang sama seperti saat Clustering
# (Pastikan df_final sudah ada dari runningan sebelumnya)
X_test = df_final[['Luas_Deforestasi', 'Jumlah_Bencana']]

# Standarisasi (Sama seperti proses sebelumnya)
scaler_test = StandardScaler()
X_test_scaled = scaler_test.fit_transform(X_test)

# Ambil Label Cluster yang sudah terbentuk
labels = df_final['Cluster']

# 2. Hitung Skor Silhouette
score = silhouette_score(X_test_scaled, labels)

print(f"\n✅ Silhouette Score Anda: {score:.4f}")

# 3. INTERPRETASI (Untuk Bab 4)
print("\n--- PANDUAN INTERPRETASI SKRIPSI ---")
if score > 0.5:
    print("Hasil: STRONG (Kuat). Struktur cluster sangat jelas terpisah.")
elif score > 0.25:
    print("Hasil: MEDIUM (Sedang). Struktur cluster ada, tapi agak berdekatan.")
else:
    print("Hasil: WEAK (Lemah). Data sangat menyatu (wajar untuk fenomena alam).")

print(f"Cara bicaranya nanti: 'Berdasarkan pengujian Silhouette Coefficient, model menghasilkan skor {score:.2f}, yang menunjukkan bahwa pengelompokan wilayah sudah terbentuk dengan cukup baik.'")
```
